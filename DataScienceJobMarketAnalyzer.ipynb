```python
## 9. Most Demanded Skills
skills = [
    "Python",
    "SQL",
    "R",
    "Excel",
    "Tableau",
    "Power BI",
    "AWS",
    "Azure",
    "Spark",
    "Hadoop",
    "TensorFlow",
    "PyTorch",
    "Machine Learning",
    "Deep Learning",
    "Statistics",
    "Pandas",
    "NumPy",
    "Scikit-learn"
]

skill_count = {}

for skill in skills:
    skill_count[skill] = df["Job Description"].str.contains(skill, case=False, na=False).sum()

skill_count
```

```python
skills_df = pd.DataFrame(
    skill_count.items(),
    columns=["Skill","Count"]
)

skills_df = skills_df.sort_values(by="Count", ascending=False)

skills_df
```

```python
plt.figure(figsize=(12,6))

plt.bar(skills_df["Skill"], skills_df["Count"])

plt.title("Most Demanded Skills")
plt.xlabel("Skills")
plt.ylabel("Number of Job Postings")

plt.xticks(rotation=60)

plt.show()
```

```python
## 10. Salary Estimate Analysis
df["Salary Estimate"].head(10)
```

```python
salary = df["Salary Estimate"].str.replace("(Glassdoor est.)", "", regex=False)
salary = salary.str.replace("$", "", regex=False)
salary = salary.str.replace("K", "", regex=False)
salary = salary.str.strip()

salary.head()
```

```python
df["Min Salary"] = salary.str.split("-").str[0]
df["Max Salary"] = salary.str.split("-").str[1]
```

```python
df["Min Salary"] = pd.to_numeric(df["Min Salary"], errors="coerce")
df["Max Salary"] = pd.to_numeric(df["Max Salary"], errors="coerce")
```

```python
df["Average Salary"] = (df["Min Salary"] + df["Max Salary"]) / 2

df[["Min Salary", "Max Salary", "Average Salary"]].head()
```

```python
plt.figure(figsize=(10,6))

plt.hist(df["Average Salary"].dropna(), bins=20)

plt.title("Average Salary Distribution")
plt.xlabel("Salary (Thousands of USD)")
plt.ylabel("Number of Jobs")

plt.show()
```

```python
highest_salary = df.groupby("Job Title")["Average Salary"].mean().sort_values(ascending=False)

highest_salary.head(10)
```

```python
plt.figure(figsize=(12,6))

highest_salary.head(10).plot(kind="bar")

plt.title("Top 10 Highest Paying Job Titles")
plt.xlabel("Job Title")
plt.ylabel("Average Salary (Thousands of USD)")

plt.xticks(rotation=60)

plt.show()
```

```python
location_salary = df.groupby("Location")["Average Salary"].mean().sort_values(ascending=False)

location_salary.head(10)
```

```python
plt.figure(figsize=(12,6))

location_salary.head(10).plot(kind="bar")

plt.title("Top Paying Locations")
plt.xlabel("Location")
plt.ylabel("Average Salary")

plt.xticks(rotation=60)

plt.show()
```
